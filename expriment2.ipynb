```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer ,WordNetLemmatizer
from nltk import pos_tag

# PorterStemmer (rute-based stemmer),
# WordNetLemmatizen (dictionary-based temmatizer), and pos_tag (assigns grammatical tags like noun/verb to each word).

text = "The children were running and playing happily in the gardens"
tokens = word_tokenize(text)

# Stemming
stemmer = PorterStemmer()
stems = [stemmer.stem(w) for w in tokens]
print("Stemmed Words:", stems)


# Lematization
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(w) for w in tokens]
print ("\nLemmatized Words:", lemmas)

# POS Togging (Penn Treebank tog set)
pos_tags = pos_tag(tokens)
print("\nPOS Tags:", pos_tags)


```
